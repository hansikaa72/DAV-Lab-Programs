```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder


data = {
    'Name': ['Anu', 'Ravi', 'Priya', 'Kiran', 'Meena', 'Arjun'],
    'Age': [21, 22, None, 23, 21, None],
    'Salary': [25000, None, 30000, 28000, None, 32000],
    'Department': ['CSE', 'ECE', 'CSE', None, 'ISE', 'ECE'],
    'City': ['Mysore', 'Bangalore', 'Mysore', 'Hubli', 'Bangalore', 'Mysore']
}

df = pd.DataFrame(data)

print("Original Data:")
print(df)



print("\nMissing Values:")
print(df.isnull())

print("\nNumber of Missing Values in Each Column:")
print(df.isnull().sum())



print("\nMean Age:", df['Age'].mean())

# Replace missing Age values with mean
df['Age'] = df['Age'].fillna(df['Age'].mean())

print("\nAfter Mean Imputation:")
print(df)



data = {
    'Name': ['Anu', 'Ravi', 'Priya', 'Kiran', 'Meena', 'Arjun'],
    'Age': [21, 22, None, 23, 21, None],
    'Salary': [25000, None, 30000, 28000, None, 32000],
    'Department': ['CSE', 'ECE', 'CSE', None, 'ISE', 'ECE'],
    'City': ['Mysore', 'Bangalore', 'Mysore', 'Hubli', 'Bangalore', 'Mysore']
}

df = pd.DataFrame(data)

print("\nMedian Salary:", df['Salary'].median())

# Replace missing Salary values with median
df['Salary'] = df['Salary'].fillna(df['Salary'].median())

print("\nAfter Median Imputation:")
print(df)


data = {
    'Student': ['Anu', 'Ravi', 'Priya', 'Kiran', 'Meena'],
    'Department': ['CSE', 'ECE', None, 'ISE', None]
}

df = pd.DataFrame(data)

print("\nBefore Forward Fill:")
print(df)

# Forward fill
df['Department'] = df['Department'].ffill()

print("\nAfter Forward Fill:")
print(df)



data = {
    'Name': ['Anu', 'Ravi', 'Priya', 'Kiran'],
    'Department': ['CSE', 'ECE', 'ISE', 'CSE']
}

df = pd.DataFrame(data)

encoder = LabelEncoder()

df['Department_Encoded'] = encoder.fit_transform(df['Department'])

print("\nLabel Encoding:")
print(df)

print("\nClasses:")
print(encoder.classes_)


data = {
    'Name': ['Anu', 'Ravi', 'Priya', 'Kiran'],
    'Department': ['CSE', 'ECE', 'ISE', 'CSE']
}

df = pd.DataFrame(data)

encoded_df = pd.get_dummies(
    df,
    columns=['Department']
)

print("\nOne-Hot Encoding:")
print(encoded_df)



encoded_df = pd.get_dummies(
    df,
    columns=['Department'],
    dtype=int
)

print("\nOne-Hot Encoding with 0 and 1:")
print(encoded_df)
```
